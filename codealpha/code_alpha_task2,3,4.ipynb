```python
import pandas as pd

url = "https://raw.githubusercontent.com/Arjun-Mota/amazon-product-reviews-sentiment-analysis/master/1429_1.csv"
df = pd.read_csv(url)


keep = ["name", "categories", "reviews.rating", "reviews.title", "reviews.text", "reviews.date", "reviews.doRecommend"]
df2 = df[keep].copy()
df2.columns = ["product_name", "category", "rating", "review_title", "review_text", "review_date", "recommend"]

df2 = df2.dropna(subset=["review_text", "rating"])
df2["rating"] = df2["rating"].astype(int)
df2 = df2.drop_duplicates(subset=["review_text"])

df2.to_csv("amazon_reviews_clean.csv", index=False)
print(df2.shape)
df2.head()
```

Question 2: Do negative reviews (1-2 stars) tend to be longer?

```python
df2['review_length'] = df2['review_text'].astype(str).str.len()
df2.groupby('rating')['review_length'].mean().round(1)
```

Result: Clear pattern - review length decreases as rating increases. 1 and
2-star reviews average ~250 characters, while 5-star reviews average only
~148 characters. This confirms the hypothesis: unhappy customers tend to
write more detailed complaints, while satisfied customers often leave short
praise ("Great!", "Love it!").

```python
from scipy import stats

# Formal hypothesis test: is review length significantly different
# between negative (1-2 star) and positive (4-5 star) reviews?
negative_lengths = df2[df2['rating'] <= 2]['review_length']
positive_lengths = df2[df2['rating'] >= 4]['review_length']

t_stat, p_value = stats.ttest_ind(negative_lengths, positive_lengths)
print(f"t-statistic: {t_stat:.2f}")
print(f"p-value: {p_value:.10f}")

if p_value < 0.05:
    print("Statistically significant difference (p < 0.05)")
else:
    print("No statistically significant difference")
```

Formal hypothesis test: Since rating and recommend are both categorical variables, I used a chi-square test of independence instead of a t-test. Result: chi2 = 18502.38, p < 0.0001. This confirms what the crosstab already suggested - there is a statistically significant association between star rating and recommend status.

Q5 : Does the average rating change over time?

```python
df2['review_date'] = pd.to_datetime(df2['review_date'], errors='coerce')
df2['year_month'] = df2['review_date'].dt.to_period('M')
df2.groupby('year_month')['rating'].mean().round(2)
```

```python
df2.groupby('year_month')['rating'].count().tail(6)
```

Result: The average rating stays fairly stable between ~4.5-4.7 from 2015
through late 2017. The apparent drop in early 2018 (down to 3.00-4.00) is
misleading - those months only have 2-3 reviews each, compared to 33-67
reviews in the months before. With such a small sample, one or two negative
reviews can swing the average dramatically. This isn't a real quality trend,
just a sample size artifact, and it's a good reminder to always check the
count alongside any average before drawing conclusions.

```python
from scipy import stats

# Formal hypothesis test: is review length significantly different
# between negative (1-2 star) and positive (4-5 star) reviews?
negative_lengths = df2[df2['rating'] <= 2]['review_length']
positive_lengths = df2[df2['rating'] >= 4]['review_length']

t_stat, p_value = stats.ttest_ind(negative_lengths, positive_lengths)
print(f"t-statistic: {t_stat:.2f}")
print(f"p-value: {p_value:.10f}")

if p_value < 0.05:
    print("Statistically significant difference (p < 0.05)")
else:
    print("No statistically significant difference")
```

Formal hypothesis test: To confirm the length difference isn't just random
noise, I ran an independent t-test comparing review length between negative
(1-2 star) and positive (4-5 star) reviews. Result: t=15.05, p<0.0001. Since
p is far below the 0.05 threshold, we can confidently say this difference is
statistically significant, not due to chance - negative reviews really are
longer on average, not just in this particular sample.

```python
from vaderSentiment.vaderSentiment import SentimentIntensityAnalyzer

analyzer = SentimentIntensityAnalyzer()

# Test on one review first, to see how it works
sample_text = df2['review_text'].iloc[0]
print(sample_text)
print(analyzer.polarity_scores(sample_text))
```

```python
# Apply VADER to every review, extract the compound score
df2['sentiment_score'] = df2['review_text'].apply(lambda x: analyzer.polarity_scores(str(x))['compound'])
df2[['review_text', 'rating', 'sentiment_score']].head()
```

```python
def classify_sentiment(score):
    if score >= 0.05:
        return 'Positive'
    elif score <= -0.05:
        return 'Negative'
    else:
        return 'Neutral'

df2['sentiment_label'] = df2['sentiment_score'].apply(classify_sentiment)
df2['sentiment_label'].value_counts()
```

```python
# Question: Does VADER's sentiment prediction agree with the actual star rating?
# Simplify rating into the same 3 categories for comparison
def rating_to_sentiment(r):
    if r >= 4:
        return 'Positive'
    elif r <= 2:
        return 'Negative'
    else:
        return 'Neutral'

df2['rating_sentiment'] = df2['rating'].apply(rating_to_sentiment)

# Compare: how often do they agree?
agreement = (df2['sentiment_label'] == df2['rating_sentiment']).mean() * 100
print(f"Agreement rate: {agreement:.1f}%")

pd.crosstab(df2['rating_sentiment'], df2['sentiment_label'])
```

```python
# Question: Is 87.4% overall accuracy misleading due to class imbalance?
# Check accuracy separately for each true category
for category in ['Negative', 'Neutral', 'Positive']:
    subset = df2[df2['rating_sentiment'] == category]
    correct = (subset['sentiment_label'] == category).sum()
    total = len(subset)
    print(f"{category}: {correct}/{total} = {correct/total*100:.1f}% correct")
```

Result: Overall agreement between VADER sentiment and star rating looks good
at 87.4%, but this is misleading due to class imbalance (92% of reviews are
positive). Breaking it down by category reveals the real picture:
- Positive: 92.2% correct - VADER handles clearly positive reviews well
- Negative: 40.8% correct - VADER misclassifies most negative reviews,
  often reading them as Positive (374 cases) or Neutral (107 cases)
- Neutral: 9.2% correct - VADER almost never predicts Neutral; it tends to
  push borderline reviews toward Positive or Negative instead

Takeaway: a lexicon-based tool like VADER is reliable for obviously positive
text, but struggles with sarcasm, mixed reviews ("great price but broke
after a week"), and mild/neutral language - it has a strong bias toward
reading things as more positive than they are, likely because product
reviews often use positive words even when describing a complaint (e.g.
